### Zadanie 5. Usługa mierzona

W chmurze wszystko jest liczone. Proszę uruchomić komórkę, a następnie kliknąć wskaźnik **RAM / Dysk** w prawym górnym rogu okna Colaba — otworzy się panel *Zasoby* (*Resources*).

````python
import time
start = time.perf_counter()
suma = sum(i * i for i in range(20_000_000))
czas = time.perf_counter() - start
print(f"Obliczenie trwało {czas:.2f} s")

CENA_ZA_GODZINE = 0.10   # umowna cena maszyny z wykładu [USD/h]
print(f"Przy cenie {CENA_ZA_GODZINE} USD/h to obliczenie kosztowałoby {czas/3600*CENA_ZA_GODZINE:.6f} USD")
!uptime
````

### Zadanie 12. Z repozytorium na maszynę w chmurze

Repozytorium jest publiczne, więc każda maszyna na świecie może pobrać jego kopię — także nasza maszyna w Colabie. Pobranie kopii nazywa się **klonowaniem** (*clone*).

Proszę upewnić się, że komórka z loginem (zadanie 8) została uruchomiona, i uruchomić poniższą.

````python
import subprocess, shutil, os

if LOGIN.startswith("tu-"):
    print("Proszę wrócić do zadania 8, wpisać login i uruchomić tamtą komórkę.")
else:
    shutil.rmtree(REPO, ignore_errors=True)
    wynik = subprocess.run(["git", "clone", f"https://github.com/{LOGIN}/{REPO}.git"], capture_output=True, text=True)
    print(wynik.stderr.strip())
    if os.path.isdir(REPO):
        print("\n=== Pliki w repozytorium ===")
        for katalog, _, pliki in os.walk(REPO):
            if ".git" in katalog: continue
            for p in pliki: print(" ", os.path.join(katalog, p))
        print("\n=== Historia zmian ===")
        print(subprocess.run(["git", "-C", REPO, "log", "--oneline"], capture_output=True, text=True).stdout)
    else:
        print("\nKlonowanie się nie powiodło. Proszę sprawdzić login i to, czy repozytorium jest publiczne.")
````

Wracamy do Colaba. Te zadania sprawdzają się **same** — po uruchomieniu komórki od razu widać, które odpowiedzi są poprawne. Można poprawiać i uruchamiać ponownie dowolną liczbę razy.

````python
# Komórka pomocnicza — proszę ją uruchomić przed zadaniami 14 i 15 (niczego nie trzeba w niej zmieniać).
import hashlib

def _skrot(pytanie, odpowiedz):
    return hashlib.sha256(f"{pytanie}|{odpowiedz.strip().lower()}".encode()).hexdigest()[:12]

def sprawdz(odpowiedzi, wzorzec):
    dobrze = 0
    for pytanie, odp in odpowiedzi.items():
        if not odp.strip():
            znak = "⬜ brak odpowiedzi"
        elif _skrot(pytanie, odp) == wzorzec[pytanie]:
            znak, dobrze = "✅", dobrze + 1
        else:
            znak = "❌ proszę spróbować jeszcze raz"
        print(f"{pytanie:<52}{odp:<10}{znak}")
    print(f"\nWynik: {dobrze} / {len(wzorzec)}")
````

### Zadanie 16. Kalkulator kosztów

Poniżej kalkulator z wykładu (slajd 23), z tymi samymi umownymi cenami. Proszę go uruchomić, a następnie — dopisując własne wywołania funkcji w komórce niżej — odpowiedzieć na pytania.

````python
CENA_VM_ZA_GODZINE = 0.10      # umowna cena maszyny wirtualnej [USD/h]
CENA_DYSKU_ZA_GB_MIES = 0.08   # umowna cena dysku [USD/GB/miesiąc]
GODZIN_W_MIESIACU = 730

def koszt_miesieczny(godziny_pracy, dysk_gb=50, rabat=0.0):
    """Koszt maszyny wirtualnej: czas pracy + dysk (płatny także, gdy maszyna jest wyłączona)."""
    return godziny_pracy*CENA_VM_ZA_GODZINE*(1 - rabat) + dysk_gb*CENA_DYSKU_ZA_GB_MIES

print("Przykład — całą dobę, na żądanie, dysk 50 GB:", koszt_miesieczny(730), "USD")
print("Przykład — 100 godzin, dysk 20 GB:           ", koszt_miesieczny(100, dysk_gb=20), "USD")
````

````python
# a) Wyłączona, ale nieusunięta maszyna — płacimy tylko za dysk 500 GB
a = koszt_miesieczny(0, dysk_gb=500)

# b) Zapomniana maszyna z dyskiem 200 GB przez cały rok
b = koszt_miesieczny(0, dysk_gb=200) * 12

# c) Pięć maszyn w pracowni: 8 h dziennie przez 22 dni
GODZINY_PRACOWNI = 8 * 22
c = 5 * koszt_miesieczny(GODZINY_PRACOWNI)

# d) Te same pięć maszyn działających całą dobę
d = 5 * koszt_miesieczny(GODZIN_W_MIESIACU)

# e) Oszczędność wariantu c względem d
e = (d - c) / d * 100

# f) Próg opłacalności: VM z rabatem 40% całą dobę vs VM na żądanie
#    (dysk jest taki sam, więc się znosi)
RABAT = 0.40
koszt_vm_z_rabatem = GODZIN_W_MIESIACU * CENA_VM_ZA_GODZINE * (1 - RABAT)
f = koszt_vm_z_rabatem / CENA_VM_ZA_GODZINE

print(f"a) Wyłączona VM, dysk 500 GB (miesiąc): {a:8.2f} USD")
print(f"b) Zapomniana VM, dysk 200 GB (rok):    {b:8.2f} USD")
print(f"c) 5 VM w pracowni (8 h × 22 dni):      {c:8.2f} USD")
print(f"d) 5 VM całą dobę:                      {d:8.2f} USD")
print(f"e) Wariant c tańszy od d o:             {e:8.1f} %")
print(f"f) Próg opłacalności zobowiązania:      {f:8.0f} h")

# Kontrola: przy f godzinach oba warianty kosztują tyle samo
assert abs(koszt_miesieczny(f) - koszt_miesieczny(GODZIN_W_MIESIACU, rabat=RABAT)) < 1e-9
````

### Zadanie 17. Elastyczność na liczbach

Sklep internetowy, jedna doba, obciążenie w kolejnych godzinach (0–23). Jedna maszyna obsługuje 10 jednostek obciążenia i kosztuje 0,10 USD za godzinę.

- **Własna serwerownia**: kupujemy tyle maszyn, ile trzeba w szczycie, i trzymamy je całą dobę.
- **Chmura**: w każdej godzinie działa tylko tyle maszyn, ile akurat trzeba.

Proszę uruchomić komórkę i odpowiedzieć na pytania 1–5.

````python
import math

obciazenie = [12, 10, 8, 8, 9, 14, 25, 38, 52, 61, 66, 70, 74, 71, 65, 60, 58, 63, 78, 92, 85, 60, 35, 20]
PROMOCJA = True          # w pytaniu 6 proszę zmienić na True

JEDNA_MASZYNA = 10        # jednostek obciążenia
CENA = 0.10               # USD za godzinę pracy jednej maszyny

maszyny_stale = math.ceil(max(obciazenie) / JEDNA_MASZYNA)     # serwerownia kupiona pod zwykły szczyt
ruch = list(obciazenie)
if PROMOCJA:              # niezapowiedziana promocja: o 19 i 20 ruch dwa razy większy
    ruch[19] *= 2; ruch[20] *= 2

maszynogodziny_stale = maszyny_stale * 24
maszynogodziny_chmura = sum(math.ceil(x / JEDNA_MASZYNA) for x in ruch)
niedobor = [g for g, x in enumerate(ruch) if x > maszyny_stale * JEDNA_MASZYNA]

print(f"Szczyt obciążenia: {max(ruch)} jednostek o godzinie {ruch.index(max(ruch))}")
print(f"Własna serwerownia: {maszyny_stale} maszyn przez całą dobę = {maszynogodziny_stale} maszynogodzin, koszt {maszynogodziny_stale*CENA:.2f} USD")
print(f"   średnie wykorzystanie: {sum(ruch)/(maszyny_stale*JEDNA_MASZYNA*24):.0%}")
print(f"   godziny, w których zabrakło mocy: {niedobor if niedobor else 'brak'}")
print(f"Chmura: {maszynogodziny_chmura} maszynogodzin, koszt {maszynogodziny_chmura*CENA:.2f} USD")
print(f"   oszczędność względem serwerowni: {1 - maszynogodziny_chmura/maszynogodziny_stale:.0%}")

import matplotlib.pyplot as plt
fig, ax = plt.subplots(figsize=(11, 3.8))
ax.bar(range(24), ruch, color="#19a7ce", alpha=.45, label="Obciążenie")
ax.axhline(maszyny_stale*JEDNA_MASZYNA, color="#b91c1c", ls="--", lw=2, label="Pojemność własnej serwerowni")
ax.step(range(24), [math.ceil(x/JEDNA_MASZYNA)*JEDNA_MASZYNA for x in ruch], where="mid", color="#0f766e", lw=2, label="Pojemność w chmurze")
ax.set_xticks(range(24)); ax.set_xlabel("Godzina"); ax.set_ylabel("Jednostki obciążenia")
ax.legend(frameon=False, loc="upper left"); ax.spines[["top", "right"]].set_visible(False)
plt.show()
````
